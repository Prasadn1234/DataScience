# Teaching Integers: A Lesson Plan & Visual Toolkit

A practical guide for introducing integers (positive, negative, and zero) to students, paired with runnable Python visualizations you can project in class.

**Grade band:** upper elementary / middle school (adapt depth as needed)

**Learning objectives**
- Understand what integers are and where they show up in real life
- Place and compare integers on a number line
- Add and subtract integers using a concrete model (chips) before moving to rules
- Recognize and correct common misconceptions

## 1. Start concrete, not with rules

Students remember "minus times minus is plus" as a rule they half-forget. They remember *owing money* or *temperature dropping below zero* because it maps to something they've lived.

Use the **Concrete → Pictorial → Abstract** progression:

1. **Concrete:** physical chips/counters (two colors: one for +1, one for -1), or acting out steps forward/backward on a taped number line on the floor.
2. **Pictorial:** number line diagrams, chip drawings (see code below).
3. **Abstract:** the symbolic rules (`-3 + 5 = 2`, etc.) — introduced *after* students can already predict the answer from the model.

### Real-world anchors to open the topic
- **Temperature:** "It was -5°C this morning and rose 8 degrees by noon. What's the temperature now?"
- **Elevation / sea level:** submarines below sea level, mountain peaks above.
- **Money:** owing money (debt = negative) vs. having money (positive).
- **Sports/game scores:** yard loss in football, points lost in a quiz game.
- **Elevators:** floors below ground level (B1, B2...) vs. above.

## 2. The Number Line

Run the cell below in class. It draws a number line and marks a value — great for a "where does this integer live?" warm-up, or for showing comparisons (`-8 < -3` because -8 is further left).

```python
import matplotlib.pyplot as plt

def draw_number_line(highlight=None, lo=-10, hi=10):
    """Draw a number line from lo to hi, optionally marking one or more integers."""
    fig, ax = plt.subplots(figsize=(10, 2))
    ax.hlines(0, lo, hi, color='black', linewidth=1)
    for x in range(lo, hi + 1):
        ax.vlines(x, -0.05, 0.05, color='black')
        ax.text(x, -0.25, str(x), ha='center', fontsize=10)
    if highlight is not None:
        pts = highlight if isinstance(highlight, (list, tuple)) else [highlight]
        for p in pts:
            ax.plot(p, 0, 'o', color='crimson', markersize=14, zorder=5)
    ax.set_xlim(lo - 1, hi + 1)
    ax.set_ylim(-0.6, 0.6)
    ax.axis('off')
    plt.show()

# Example: compare -8 and -3
draw_number_line(highlight=[-8, -3])
```

**In-class prompt:** "Which is bigger, -8 or -3? Point to it on the line." Students often guess -8 because 8 > 3 as whole numbers — this is the single most common misconception (see section 5). The number line makes the direction visual: further right = larger.

## 3. Addition & Subtraction with a Chip Model

**Zero pair rule:** one positive chip (+1) and one negative chip (-1) cancel out to zero.

- To add integers, combine chips and cancel zero pairs; what's left is the answer.
- To subtract, students *remove* chips (adding zero pairs first if you don't have enough of the right color to remove).

Example: `-3 + 5`
1. Lay out 3 negative chips and 5 positive chips.
2. Cancel 3 zero pairs (3 negatives with 3 positives).
3. Left with 2 positive chips → answer is `2`.

The function below prints a simple ASCII chip diagram so you can generate examples on the fly.

```python
def chip_model(a, b):
    """Show the zero-pair chip model for a + b."""
    pos = ['+'] * max(a, 0) + ['+'] * max(b, 0)
    neg = ['-'] * max(-a, 0) + ['-'] * max(-b, 0)
    cancel = min(len(pos), len(neg))
    print(f"{a} + {b}")
    print("Before:  ", ' '.join(pos), '|', ' '.join(neg))
    print(f"Cancel {cancel} zero pair(s)...")
    remaining_pos = pos[cancel:]
    remaining_neg = neg[cancel:]
    result = len(remaining_pos) - len(remaining_neg)
    print("After:   ", ' '.join(remaining_pos) if remaining_pos else ' '.join(remaining_neg))
    print(f"Answer: {result}")

chip_model(-3, 5)
print()
chip_model(4, -7)
```

**Subtraction tip:** teach "subtracting is adding the opposite" (`a - b = a + (-b)`) using the same chips, so students only ever need one model for both operations.

## 4. Suggested Lesson Sequence (5 days)

| Day | Focus | Activity |
|---|---|---|
| 1 | What are integers? | Sort real-world cards (temps, elevations, debts) into positive/negative/zero. Introduce vocabulary. |
| 2 | Number line & comparing | Human number line (tape on floor, students stand on values). Compare/order integers. |
| 3 | Adding integers | Chip model with physical counters, then pictorial (draw chips), then a few abstract problems. |
| 4 | Subtracting integers | "Subtract = add the opposite." Practice converting problems, then solve with chips. |
| 5 | Practice & real problems | Word problems (temperature swings, football yardage, bank balance). Exit ticket / quiz. |

## 5. Common Misconceptions to Watch For

- **"-8 is bigger than -3 because 8 > 3."** Fix with the number line — direction, not digit size, determines order.
- **Confusing the negative sign with subtraction.** Reinforce that `-5` is a value (five below zero), while `7 - 5` is an operation. Reading `-5` as "negative five" (not "minus five") helps.
- **"Adding always makes bigger, subtracting always makes smaller."** Counterexample: `-2 + (-3) = -5` is smaller than -2; `-2 - (-5) = 3` is bigger than -2. Use real contexts (temperature drop is "adding a negative") to break this.
- **Rushing to symbolic rules.** If a student can recite "same signs add and keep the sign, different signs subtract" but can't explain why with chips or a number line, they don't understand it yet — they'll misapply it under pressure.

## 6. Quick Formative Check (Exit Ticket Ideas)

1. Order these from least to greatest: `-7, 3, -2, 0, -10`
2. The temperature was -4°C and dropped 6 more degrees. What is it now?
3. Draw a chip model for `-6 + 2` and give the answer.
4. True or false: `-15 < -20`. Explain using the number line.
